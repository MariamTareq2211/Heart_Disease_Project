```python
# 07_final_model_export.ipynb

import pandas as pd
import joblib
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, roc_auc_score
from sklearn.ensemble import RandomForestClassifier

# -----------------------------
# Load reduced dataset
# -----------------------------
df = pd.read_csv("../data/heart_disease_reduced.csv")

X = df.drop(columns=["target"])
y = df["target"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# -----------------------------
# Load Best Model (example: Random Forest)
# -----------------------------
# NOTE: Replace with the best model you found in Notebook 6
best_model = RandomForestClassifier(
    n_estimators=200, max_depth=8, min_samples_split=5,
    criterion="gini", random_state=42
)

best_model.fit(X_train, y_train)

# -----------------------------
# Evaluate final model
# -----------------------------
y_pred = best_model.predict(X_test)
y_proba = best_model.predict_proba(X_test)[:, 1]

report = classification_report(y_test, y_pred, output_dict=True)
roc_auc = roc_auc_score(y_test, y_proba)

# Save evaluation metrics
results_path = "../results/evaluation_metrics.txt"
with open(results_path, "w") as f:
    f.write("Final Model: Random Forest\n")
    f.write(f"ROC AUC: {roc_auc:.4f}\n\n")
    f.write("Classification Report:\n")
    for label, metrics in report.items():
        f.write(f"{label}: {metrics}\n")

print(f"✅ Evaluation metrics saved to {results_path}")

# -----------------------------
# Save trained model
# -----------------------------
model_path = "../models/final_model.pkl"
joblib.dump(best_model, model_path)
print(f"✅ Final model saved to {model_path}")
```
